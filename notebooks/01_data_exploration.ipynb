```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

# Plot styling
sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["figure.dpi"] = 100

# Project paths
DATA_DIR = Path("../data")
RAW_DIR = DATA_DIR / "raw"
PROCESSED_DIR = DATA_DIR / "processed"
FIGURES_DIR = Path("../figures/exploration")

# Create output directories if they don't exist
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
```

```python
hour = pd.read_csv(RAW_DIR / "hour.csv")
day = pd.read_csv(RAW_DIR / "day.csv")

print("Hourly dataset:", hour.shape)
print("Daily dataset:", day.shape)

display(hour.head())
display(day.head())
```

```python
print("HOURLY DATASET")
hour.info()

print("\nDAILY DATASET")
day.info()
```

```python
display(hour.describe().T)
display(day.describe().T)
```

```python
def data_quality_report(df, name):
    print(f"\n--- {name} ---")

    print("Rows:", len(df))
    print("Columns:", len(df.columns))

    print("\nMissing values:")
    print(df.isnull().sum()[df.isnull().sum() > 0])

    print("\nDuplicate rows:", df.duplicated().sum())

    print("\nDate range:")
    print(df["dteday"].min(), "to", df["dteday"].max())

    print("\nNegative rental counts:")
    print((df["cnt"] < 0).sum())

data_quality_report(hour, "Hourly Dataset")
data_quality_report(day, "Daily Dataset")
```

```python
# Make copies so raw data remains unchanged
hour_clean = hour.copy()
day_clean = day.copy()

# Convert dates
hour_clean["dteday"] = pd.to_datetime(hour_clean["dteday"])
day_clean["dteday"] = pd.to_datetime(day_clean["dteday"])

# Sort chronologically
hour_clean = hour_clean.sort_values(
    ["dteday", "hr"]
).reset_index(drop=True)

day_clean = day_clean.sort_values(
    "dteday"
).reset_index(drop=True)

# Convert categorical codes into readable labels
season_labels = {
    1: "Spring",
    2: "Summer",
    3: "Fall",
    4: "Winter"
}

weather_labels = {
    1: "Clear / Mostly Clear",
    2: "Mist / Cloudy",
    3: "Light Snow / Rain",
    4: "Heavy Rain / Snow"
}

for df in [hour_clean, day_clean]:
    df["season_name"] = df["season"].map(season_labels)
    df["weather_name"] = df["weathersit"].map(weather_labels)

    df["month_name"] = df["mnth"].map({
        i: pd.Timestamp(2020, i, 1).strftime("%B")
        for i in range(1, 13)
    })

# Save cleaned datasets
hour_clean.to_csv(
    PROCESSED_DIR / "hour_clean.csv", index=False
)

day_clean.to_csv(
    PROCESSED_DIR / "day_clean.csv", index=False
)

print("Cleaned datasets saved successfully.")
```

```python
comparison = pd.DataFrame({
    "Hourly": [
        len(hour_clean),
        hour_clean["dteday"].nunique(),
        hour_clean["cnt"].mean(),
        hour_clean["cnt"].std()
    ],
    "Daily": [
        len(day_clean),
        day_clean["dteday"].nunique(),
        day_clean["cnt"].mean(),
        day_clean["cnt"].std()
    ]
}, index=[
    "Number of observations",
    "Unique dates",
    "Average rentals per observation",
    "Standard deviation of rentals"
])

display(comparison)
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    day_clean["dteday"],
    day_clean["cnt"],
    alpha=0.35,
    label="Daily rentals"
)

# 30-day rolling average
rolling_avg = day_clean["cnt"].rolling(
    window=30,
    center=True
).mean()

plt.plot(
    day_clean["dteday"],
    rolling_avg,
    linewidth=2.5,
    label="30-day rolling average"
)

plt.title("Bike Rental Demand Over Time", fontsize=16)
plt.xlabel("Date")
plt.ylabel("Number of Rentals")
plt.legend()
plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "01_rentals_over_time.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

as expected, demand is lower in the winter and higher over the summer.

```python
monthly_demand = (
    day_clean.groupby("mnth")["cnt"]
    .mean()
    .reset_index()
)

month_labels = [
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug",
    "Sep", "Oct", "Nov", "Dec"
]

plt.figure(figsize=(12, 6))

sns.barplot(
    data=monthly_demand,
    x="mnth",
    y="cnt"
)

plt.xticks(range(12), month_labels)
plt.title("Average Daily Bike Rentals by Month")
plt.xlabel("Month")
plt.ylabel("Average Daily Rentals")
plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "02_rentals_by_month.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

as expected, summer months have higher rentals compared to winter months as its presumably colder.

```python
season_order = [
    "Spring",
    "Summer",
    "Fall",
    "Winter"
]

plt.figure(figsize=(10, 6))

sns.barplot(
    data=day_clean,
    x="season_name",
    y="cnt",
    order=season_order,
    errorbar=None
)

plt.title("Average Daily Bike Rentals by Season")
plt.xlabel("Season")
plt.ylabel("Average Daily Rentals")
plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "03_rentals_by_season.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

note that spring is considered jan-mar and winter is considered oct-dec, so nicer days in october can lead to higher numbers in winter

```python
hourly_demand = (
    hour_clean.groupby("hr")["cnt"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(13, 6))

sns.lineplot(
    data=hourly_demand,
    x="hr",
    y="cnt",
    marker="o",
    linewidth=2
)

plt.xticks(range(24))
plt.title("Average Bike Rentals by Hour of Day")
plt.xlabel("Hour of Day (0–23)")
plt.ylabel("Average Rentals per Hour")
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "04_rentals_by_hour.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

peak commuting hours are 8am and 5pm (followed closely by 6pm), this corresponds to very standard corporate job 9-5 work days

```python
plt.figure(figsize=(12, 6))

sns.histplot(
    day_clean["cnt"],
    bins=30,
    kde=True
)

plt.title("Distribution of Daily Bike Rentals")
plt.xlabel("Number of Rentals per Day")
plt.ylabel("Frequency")
plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "05_daily_rental_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

almost a typical bell curve but then has mini peaks around 1500 and 7500, which is kinda odd. maybe can explain the higher number by a lot more casual riders renting bikes on certain days (probably in the summer). could also explain the 1500 number by regulars who ride no matter what / have no alternate mode of transport

```python
plt.figure(figsize=(12, 6))

sns.histplot(
    hour_clean["cnt"],
    bins=40,
    kde=True
)

plt.title("Distribution of Hourly Bike Rentals")
plt.xlabel("Number of Rentals per Hour")
plt.ylabel("Frequency")
plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "06_hourly_rental_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
